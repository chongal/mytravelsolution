````python
!pip uninstall instagrapi -y
!pip install instagrapi==1.17.1
from instagrapi import Client
from datetime import datetime, timedelta, timezone
import csv
import time

# === Авторизация ===
cl = Client()

# ВАРИАНТ 1: Логин через username/password (РЕКОМЕНДУЕТСЯ)
# cl.login("ваш_username", "ваш_password")

# ВАРИАНТ 2: Загрузка сохраненной сессии
import os
session_file = "session.json"

try:
    if os.path.exists(session_file):
        print("🔄 Загружаем сохраненную сессию...")
        cl.load_settings(session_file)
        cl.login("ваш_username", "ваш_password")  # переавторизуемся с сохраненными настройками
        print("✅ Сессия загружена успешно!")
    else:
        print("🔑 Первый вход - создаем новую сессию...")
        cl.login("ваш_username", "ваш_password")
        cl.dump_settings(session_file)
        print("✅ Сессия сохранена!")
except Exception as e:
    print(f"❌ Ошибка авторизации: {e}")
    print("\n💡 Возможные решения:")
    print("   1. Проверьте логин и пароль")
    print("   2. Instagram может требовать подтверждение через приложение")
    print("   3. Включите двухфакторную аутентификацию и используйте код")
    exit()

# === Настройки ===
username = "tau_dastarkhan"
output_file = "tau_posts.csv"
days_back = 30
# Используем timezone-aware дату
cutoff_date = datetime.now(timezone.utc) - timedelta(days=days_back)

# === Получаем инфо о пользователе ===
try:
    user = cl.user_info_by_username(username)
    print(f"✅ Пользователь найден: {user.username}")
    print(f"📊 Подписчиков: {user.follower_count}")
    print(f"📸 Всего постов: {user.media_count}")
    print(f"📝 Bio: {user.biography}\n")
except Exception as e:
    print("❌ Ошибка при получении профиля:", e)
    exit()

# === Получаем ВСЕ посты с правильной пагинацией ===
print("📥 Загружаем посты за последний месяц...\n")
user_id = user.pk
all_posts = []
end_cursor = None
page = 1
stop_parsing = False

try:
    while not stop_parsing:
        print(f"🔄 Загрузка страницы {page}...")

        # Запрашиваем посты пачками по 50
        if end_cursor:
            medias = cl.user_medias_v1(user_id, amount=50, end_cursor=end_cursor)
        else:
            medias = cl.user_medias_v1(user_id, amount=50)

        if not medias:
            print("✅ Все посты загружены!")
            break

        # Обрабатываем каждый пост
        posts_in_range = 0
        for post in medias:
            post_date = post.taken_at

            # Проверяем дату (делаем обе даты timezone-aware)
            if not post_date.tzinfo:
                post_date = post_date.replace(tzinfo=timezone.utc)

            # Извлекаем URL медиа
            if post.media_type == 1:  # фото
                media_url = post.thumbnail_url
            elif post.media_type == 2:  # видео
                media_url = post.video_url
            elif post.media_type == 8:  # карусель
                media_url = post.resources[0].thumbnail_url if post.resources else None
            else:
                media_url = None

            # Проверяем, входит ли пост в нужный диапазон
            if post_date >= cutoff_date:
                # Формируем правильную ссылку на пост
                post_code = getattr(post, 'code', None)
                if not post_code and hasattr(post, 'pk'):
                    # Если code нет, используем ID поста
                    post_url = f"https://www.instagram.com/p/{post.pk}/"
                elif post_code:
                    post_url = f"https://www.instagram.com/p/{post_code}/"
                else:
                    post_url = ""

                all_posts.append({
                    "date": post_date.strftime("%Y-%m-%d %H:%M"),
                    "type": "photo" if post.media_type == 1 else "video" if post.media_type == 2 else "carousel",
                    "caption": (post.caption_text or "")[:200],  # первые 200 символов
                    "likes": post.like_count,
                    "comments": post.comment_count,
                    "url": media_url or "",
                    "post_url": post_url,
                    "post_id": str(post.pk)
                })
                posts_in_range += 1
                print(f"  ✓ Пост {len(all_posts)}: {post_date.strftime('%Y-%m-%d')} | {post.like_count} лайков")
            else:
                print(f"  ⏭️ Пропускаем: {post_date.strftime('%Y-%m-%d')} (старше {days_back} дней)")
                # Если встретили старый пост, останавливаем загрузку
                if posts_in_range > 0:  # но только если уже нашли хоть один подходящий
                    stop_parsing = True
                    break

        # Если есть еще посты, берем курсор для следующей страницы
        if len(medias) < 50 or stop_parsing:
            break

        # Получаем ID последнего поста для пагинации
        end_cursor = medias[-1].pk
        page += 1

        # ВАЖНО: делаем паузу, чтобы не забанили
        time.sleep(3)

except Exception as e:
    print(f"\n❌ Ошибка: {e}")
    print("💡 Попробуйте увеличить задержку или использовать другой метод")

print(f"\n🎉 Всего найдено: {len(all_posts)} постов за последние {days_back} дней")
print(f"📅 Диапазон дат: с {cutoff_date.strftime('%Y-%m-%d')} по сегодня\n")

# === Сохраняем в CSV ===
output_file = "posts.csv"

if all_posts:
    with open(output_file, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(
            f,
            fieldnames=[
                "date",
                "type",
                "caption",
                "likes",
                "comments",
                "url",
                "post_url",
                "post_id"  # добавили сюда
            ]
        )
        writer.writeheader()
        writer.writerows(all_posts)
    print(f"✅ Данные сохранены в {output_file}")
    print(f"📊 Статистика:")
    print(f"   • Всего лайков: {sum(p['likes'] for p in all_posts)}")
    print(f"   • Всего комментариев: {sum(p['comments'] for p in all_posts)}")
    print(f"   • Средний охват: {sum(p['likes'] for p in all_posts) // len(all_posts)} лайков/пост")
else:
    print("⚠️ Нет постов за указанный период.")
````

````python
import requests
from bs4 import BeautifulSoup
import ollama
import json

class WebsiteSummarizer:
    def __init__(self, model_name="llama3.2"):
        """
        Инициализация с моделью Ollama
        Рекомендуемые модели: llama3.2, mistral, qwen2.5
        """
        self.model_name = model_name

    def fetch_website(self, url):
        """Загружает содержимое веб-сайта"""
        try:
            headers = {
                'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
            }
            response = requests.get(url, headers=headers, timeout=10)
            response.raise_for_status()
            return response.text
        except Exception as e:
            return f"Ошибка при загрузке сайта: {str(e)}"

    def extract_text(self, html_content):
        """Извлекает текст из HTML"""
        soup = BeautifulSoup(html_content, 'html.parser')

        # Удаляем ненужные элементы
        for tag in soup(['script', 'style', 'nav', 'footer', 'header']):
            tag.decompose()

        # Извлекаем текст
        text = soup.get_text(separator=' ', strip=True)

        # Очищаем от лишних пробелов
        text = ' '.join(text.split())

        # Ограничиваем длину для оптимальной работы LLM
        return text[:8000]

    def create_summary(self, url, max_sentences=6, language="ru"):
        """Создает summary сайта"""

        # 1. Загружаем сайт
        print(f"Загрузка сайта: {url}")
        html_content = self.fetch_website(url)

        if html_content.startswith("Ошибка"):
            return html_content

        # 2. Извлекаем текст
        print("Извлечение текста...")
        text_content = self.extract_text(html_content)

        # 3. Создаем промпт для Ollama
        prompt = f"""Проанализируй следующий текст с веб-сайта зоны отдыха или туристического комплекса.

Текст сайта:
{text_content}

Задание: Создай краткое описание (summary) этого места для отдыха. Требования:
- Максимум {max_sentences} предложений
- Язык: {"русский" if language == "ru" else "английский"}
- Укажи основные удобства и услуги
- Упомяни местоположение, если есть информация
- Сделай текст привлекательным и информативным
- Не добавляй заголовки, только текст summary

Summary:"""

        # 4. Отправляем в Ollama
        print(f"Генерация summary с моделью {self.model_name}...")
        try:
            response = ollama.generate(
                model=self.model_name,
                prompt=prompt,
                options={
                    'temperature': 0.7,
                    'top_p': 0.9,
                    'num_predict': 600
                }
            )

            summary = response['response'].strip()
            return summary

        except Exception as e:
            return f"Ошибка при генерации summary: {str(e)}"

    def batch_summarize(self, urls, max_sentences=6):
        """Создает summary для нескольких URL"""
        results = {}
        for url in urls:
            print(f"\n{'='*60}")
            summary = self.create_summary(url, max_sentences)
            results[url] = summary
            print(f"\nSummary для {url}:")
            print(summary)
        return results


# Пример использования
if __name__ == "__main__":
    # Инициализация
    summarizer = WebsiteSummarizer(model_name="qwen3:14b")

    # Один сайт
    url = "https://fachwerk-borovoe.kz"
    summary = summarizer.create_summary(url, max_sentences=6, language="ru")
    print("\n" + "="*60)
    print("РЕЗУЛЬТАТ:")
    print("="*60)
    print(summary)

    # Несколько сайтов (раскомментируйте при необходимости)
    # urls = [
    #     "https://tau24.kz",
    #     "https://example-resort.com"
    # ]
    # results = summarizer.batch_summarize(urls, max_sentences=6)

    # Сохранение в JSON
    # with open('summaries.json', 'w', encoding='utf-8') as f:
    #     json.dump(results, f, ensure_ascii=False, indent=2)
````

````python
import requests
from bs4 import BeautifulSoup
import ollama
import json
import re

class WebsiteSummarizer:
    def __init__(self, model_name="qwen3:14b"):
        """
        Инициализация с моделью Ollama
        Рекомендуемые модели: qwen3:14b, llama3.2, mistral, qwen2.5
        """
        self.model_name = model_name

    def _extract_json_fields(self, text):
        """Извлекает поля из поврежденного JSON или текста"""
        info = {
            "name": "неизвестно",
            "type": "место отдыха",
            "location": "неизвестно",
            "social_media": "неизвестно",
            "key_features": "неизвестно"
        }

        print(f"\n🔧 Fallback: извлечение через regex")
        print(f"Текст для анализа (первые 300 символов):\n{text[:300]}\n")

        # Извлекаем каждое поле отдельно с несколькими вариантами паттернов
        patterns = {
            "name": [
                r'"name"\s*:\s*"([^"\\]*(?:\\.[^"\\]*)*)"',
                r'"name"\s*:\s*"([^"]+?)(?:"|,|\})',
                r'"name":\s*"([^"]+)"',
                r'name["\s:]+([А-Яа-яA-Za-z0-9\s\-]+?)(?:"|,)'
            ],
            "type": [
                r'"type"\s*:\s*"([^"\\]*(?:\\.[^"\\]*)*)"',
                r'"type"\s*:\s*"([^"]+?)(?:"|,|\})',
                r'"type":\s*"([^"]+)"',
                r'type["\s:]+([А-Яа-яA-Za-z\s\-]+?)(?:"|,)'
            ],
            "location": [
                r'"location"\s*:\s*"([^"\\]*(?:\\.[^"\\]*)*)"',
                r'"location"\s*:\s*"([^"]+?)(?:"|,|\})',
                r'"location":\s*"([^"]+)"',
                r'location["\s:]+([А-Яа-яA-Za-z\s\-]+?)(?:"|,)'
            ],
            "social_media": [
                r'"social_media"\s*:\s*"([^"\\]*(?:\\.[^"\\]*)*)"',
                r'"social_media"\s*:\s*"([^"]+?)(?:"|,|\})',
                r'"social_media":\s*"([^"]+)"',
            ],
            "key_features": [
                r'"key_features"\s*:\s*"([^"\\]*(?:\\.[^"\\]*)*)"',
                r'"key_features"\s*:\s*"([^"]+?)(?:"|,|\})',
                r'"key_features":\s*"([^"]+)"',
            ]
        }

        for key, pattern_list in patterns.items():
            for i, pattern in enumerate(pattern_list):
                match = re.search(pattern, text, re.DOTALL | re.IGNORECASE)
                if match:
                    value = match.group(1).strip()
                    # Очищаем от переносов строк и лишних пробелов
                    value = ' '.join(value.split())
                    # Убираем escape символы
                    value = value.replace('\\n', ' ').replace('\\r', '').replace('\\"', '"')
                    if value and value != '' and len(value) > 2:
                        info[key] = value
                        print(f"  ✓ {key}: {value} (паттерн #{i+1})")
                        break

        return info

    def _smart_extract_from_text(self, text, title, llm_response=""):
        """Умное извлечение информации напрямую из текста сайта"""
        print(f"\n🧠 Умное извлечение из исходного текста...")

        info = {
            "name": "неизвестно",
            "type": "место отдыха",
            "location": "неизвестно",
            "social_media": "неизвестно",
            "key_features": "неизвестно"
        }

        # Извлекаем название из title
        if title and len(title) > 3:
            # Убираем общие слова из title
            clean_title = re.sub(r'\s*[-|—]\s*.*, ', title)  # Убираем все после - или |
            clean_title = clean_title.strip()
            if len(clean_title) > 3:
                info["name"] = clean_title
                print(f"  ✓ name: {clean_title} (из title)")

        # Ищем тип заведения
        type_patterns = [
            (r'(?:гостиница|hotel)', 'отель'),
            (r'(?:курорт|resort)', 'курорт'),
            (r'(?:глэмпинг|glamping)', 'глэмпинг'),
            (r'(?:база отдыха|recreation)', 'база отдыха'),
            (r'(?:санаторий|sanatorium)', 'санаторий'),
            (r'(?:спа|spa)', 'спа-центр'),
            (r'(?:коттедж|cottage)', 'коттеджный комплекс'),
            (r'(?:усадьба|estate)', 'усадьба'),
        ]


    def _translate_type_to_russian(self, type_text):
        """Переводит тип места на русский язык"""
        translations = {
            # Английские варианты
            'resort': 'курорт',
            'hotel': 'отель',
            'glamping': 'глэмпинг',
            'camping': 'кемпинг',
            'cottage': 'коттедж',
            'guesthouse': 'гостевой дом',
            'hostel': 'хостел',
            'motel': 'мотель',
            'villa': 'вилла',
            'apartment': 'апартаменты',
            'recreation center': 'база отдыха',
            'recreation complex': 'комплекс отдыха',
            'spa resort': 'спа-курорт',
            'wellness resort': 'оздоровительный курорт',
            'mountain resort': 'горный курорт',
            'eco-resort': 'эко-курорт',
            'family resort': 'семейный курорт',
            'sanatorium': 'санаторий',
            'rest house': 'дом отдыха',
            'tourist base': 'турбаза',
            'lodge': 'усадьба',
            'chalet': 'шале',
            'bungalow': 'бунгало',
            'cabin': 'домик',

            # Смешанные варианты
            'spa center': 'спа-центр',
            'wellness center': 'оздоровительный центр',
            'recreation area': 'зона отдыха',
            'holiday resort': 'курорт',
            'vacation resort': 'курорт',
        }

        # Приводим к нижнему регистру для поиска
        type_lower = type_text.lower().strip()

        # Проверяем точное совпадение
        if type_lower in translations:
            return translations[type_lower]

        # Проверяем частичное совпадение
        for eng, rus in translations.items():
            if eng in type_lower:
                return rus

        # Если уже на русском или не нашли перевод, возвращаем как есть
        return type_text

    def fetch_website(self, url):
        """Загружает содержимое веб-сайта"""
        try:
            headers = {
                'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
            }
            response = requests.get(url, headers=headers, timeout=10)
            response.raise_for_status()
            return response.text
        except Exception as e:
            return f"Ошибка при загрузке сайта: {str(e)}"

    def extract_text(self, html_content):
        """Извлекает текст из HTML"""
        soup = BeautifulSoup(html_content, 'html.parser')

        # Пытаемся найти название из title или h1
        title = ""
        if soup.title:
            title = soup.title.string

        # Удаляем ненужные элементы
        for tag in soup(['script', 'style', 'nav', 'footer', 'header']):
            tag.decompose()

        # Извлекаем текст
        text = soup.get_text(separator=' ', strip=True)

        # Очищаем от лишних пробелов
        text = ' '.join(text.split())

        # Ограничиваем длину для оптимальной работы LLM
        return text[:8000], title

    def extract_business_info(self, url, additional_urls=None):
        """Извлекает информацию о бизнесе для персонализации

        Args:
            url: основной URL сайта
            additional_urls: список дополнительных URL (например, 2ГИС, социальные сети)
        """

        # 1. Загружаем основной сайт
        print(f"Загрузка основного сайта: {url}")
        html_content = self.fetch_website(url)

        if html_content.startswith("Ошибка"):
            return {"error": html_content}

        # 2. Извлекаем текст с основного сайта
        print("Извлечение информации с основного сайта...")
        text_content, title = self.extract_text(html_content)

        # 3. Загружаем дополнительные источники
        additional_text = ""
        if additional_urls:
            print(f"\nЗагрузка дополнительных источников ({len(additional_urls)})...")
            for i, add_url in enumerate(additional_urls, 1):
                print(f"  {i}. {add_url}")
                add_html = self.fetch_website(add_url)
                if not add_html.startswith("Ошибка"):
                    add_text, _ = self.extract_text(add_html)
                    additional_text += f"\n\n--- Источник {i} ({add_url}) ---\n{add_text[:3000]}"

        # Объединяем всю информацию
        combined_text = text_content + additional_text

        print(f"\n📝 Общий объем текста для анализа: {len(combined_text)} символов")
        print(f"   - Основной сайт: {len(text_content)} символов")
        if additional_text:
            print(f"   - Доп. источники: {len(additional_text)} символов")

        # 4. Создаем улучшенный промпт с примерами
        prompt = f"""Extract business information from the text and format as JSON.

Text to analyze:
{combined_text[:10000]}

Extract these fields:
- name: business name
- type: type of establishment (hotel, resort, glamping, etc.)
- location: city or region
- social_media: social media platform mentioned
- key_features: main amenities

Respond ONLY with JSON in this exact format:
{{"name": "extracted name", "type": "extracted type", "location": "extracted location", "social_media": "extracted platform or неизвестно", "key_features": "extracted features"}}

JSON:"""

        # 5. Отправляем в Ollama с adjusted параметрами
        print(f"\nАнализ данных с моделью {self.model_name}...")

        # Пробуем несколько раз с разными настройками если не получается
        max_attempts = 2
        for attempt in range(max_attempts):
            try:
                if attempt > 0:
                    print(f"\n🔄 Попытка #{attempt + 1}...")

                response = ollama.generate(
                    model=self.model_name,
                    prompt=prompt,
                    options={
                        'temperature': 0.1 if attempt == 0 else 0.3,
                        'top_p': 0.9,
                        'num_predict': 600,
                    }
                )

                # Извлекаем ответ
                response_text = response['response'].strip()

                if attempt == 0:  # Показываем только в первый раз
                    print(f"\n🔍 Ответ модели (первые 500 символов):")
                    print("=" * 60)
                    print(response_text[:500])
                    print("=" * 60)

                # Очищаем ответ
                response_text = re.sub(r'```json\s*', '', response_text)
                response_text = re.sub(r'```\s*', '', response_text)
                response_text = response_text.strip()

                # Ищем JSON
                json_match = re.search(r'\{[^\{\}]*\}', response_text)

                if json_match:
                    json_str = json_match.group()

                    # Пытаемся распарсить
                    try:
                        parsed = json.loads(json_str)
                        # Проверяем что есть хотя бы одно нужное поле
                        if any(key in parsed for key in ["name", "type", "location"]):
                            print(f"✅ JSON успешно распарсен!")

                            # Дополняем недостающие поля
                            info = {
                                "name": parsed.get("name", "неизвестно"),
                                "type": parsed.get("type", "место отдыха"),
                                "location": parsed.get("location", "неизвестно"),
                                "social_media": parsed.get("social_media", "неизвестно"),
                                "key_features": parsed.get("key_features", "неизвестно")
                            }

                            print(f"\n📊 Извлеченные данные:")
                            for key, value in info.items():
                                print(f"  • {key}: {value}")

                            return info
                    except json.JSONDecodeError:
                        pass

                # Если JSON не найден или не распарсился, пробуем fallback
                if attempt == max_attempts - 1:
                    print(f"\n⚠️ JSON не найден или поврежден, использую умное извлечение...")
                    info = self._smart_extract_from_text(combined_text, title, response_text)

                    print(f"\n📊 Извлеченные данные:")
                    for key, value in info.items():
                        print(f"  • {key}: {value}")

                    return info

            except Exception as e:
                print(f"❌ Ошибка в попытке #{attempt + 1}: {str(e)}")
                if attempt == max_attempts - 1:
                    print(f"\n🔧 Использую аварийное извлечение...")
                    return self._smart_extract_from_text(combined_text, title, "")

    def generate_outreach_message(self, url, additional_urls=None, template=None):
        """Генерирует персонализированное сообщение рассылки

        Args:
            url: основной URL сайта
            additional_urls: список дополнительных URL (2ГИС, соцсети и т.д.)
            template: шаблон сообщения
        """

        if template is None:
            template = """Привет, {name}!

Мы заметили ваш потрясающий {type} рядом с {location} в {social_media}.
Ваше место явно заслуживает большей аудитории туристов!
mytravel.kz — это топовая платформа путешествий в Казахстане, где вы получите:
✨ 10K+ активных туристов ежемесячно
✨ Лучший рейтинг для премиум объектов
✨ Инструменты управления бронированиями

Хотим добавить вас в каталог. Можно поговорим?

Спасибо,
Команда mytravel.kz"""

        # Извлекаем информацию из всех источников
        info = self.extract_business_info(url, additional_urls)

        if "error" in info:
            return info["error"]

        # Переводим тип на русский
        type_translated = self._translate_type_to_russian(info.get("type", "место отдыха"))

        # Заполняем шаблон
        message = template.format(
            name=info.get("name", "[Название]"),
            type=type_translated,
            location=info.get("location", "[Локация]"),
            social_media=info.get("social_media", "социальных сетях")
        )

        # Сохраняем переведенный тип в info
        info["type_translated"] = type_translated

        return {
            "message": message,
            "extracted_info": info
        }

    def batch_generate_messages(self, urls_with_sources, template=None, output_file="outreach_messages.json"):
        """Генерирует сообщения для списка URL с дополнительными источниками

        Args:
            urls_with_sources: список словарей вида:
                [
                    {
                        "main_url": "https://site.com",
                        "additional_urls": ["https://2gis.kz/...", "https://instagram.com/..."]
                    },
                    ...
                ]
                или просто список URL строк (старый формат)
            template: шаблон сообщения
            output_file: имя файла для сохранения результатов
        """
        results = []

        for i, url_data in enumerate(urls_with_sources, 1):
            print(f"\n{'='*60}")

            # Поддержка старого формата (просто строка URL)
            if isinstance(url_data, str):
                main_url = url_data
                additional_urls = None
                print(f"Обработка {i}/{len(urls_with_sources)}: {main_url}")
            else:
                main_url = url_data.get("main_url")
                additional_urls = url_data.get("additional_urls", [])
                print(f"Обработка {i}/{len(urls_with_sources)}: {main_url}")
                if additional_urls:
                    print(f"  + {len(additional_urls)} дополнительных источника(ов)")

            print('='*60)

            result = self.generate_outreach_message(main_url, additional_urls, template)

            if isinstance(result, dict) and "message" in result:
                results.append({
                    "main_url": main_url,
                    "additional_urls": additional_urls if additional_urls else [],
                    "message": result["message"],
                    "info": result["extracted_info"]
                })

                print("\n📧 СГЕНЕРИРОВАННОЕ СООБЩЕНИЕ:")
                print("-" * 60)
                print(result["message"])
                print("-" * 60)
                print(f"\n📊 Извлеченная информация:")
                for key, value in result["extracted_info"].items():
                    print(f"  • {key}: {value}")
            else:
                results.append({
                    "main_url": main_url,
                    "additional_urls": additional_urls if additional_urls else [],
                    "error": result
                })
                print(f"\n❌ Ошибка: {result}")

        # Сохраняем результаты
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(results, f, ensure_ascii=False, indent=2)

        print(f"\n✅ Результаты сохранены в {output_file}")
        return results


# Пример использования
if __name__ == "__main__":
    # Инициализация
    summarizer = WebsiteSummarizer(model_name="qwen3:14b")

    # ПРИМЕР 1: Один сайт с дополнительными источниками
    print("🚀 ГЕНЕРАЦИЯ ПЕРСОНАЛИЗИРОВАННОГО СООБЩЕНИЯ\n")

    main_url = "https://tau24.kz"
    additional_urls = [
        "https://2gis.kz/almaty/firm/70000001018489394",  # 2ГИС
        # "https://instagram.com/tau_dastarkhan",  # Instagram (опционально)
    ]

    result = summarizer.generate_outreach_message(
        url=main_url,
        additional_urls=additional_urls
    )

    if isinstance(result, dict) and "message" in result:
        print("\n" + "="*60)
        print("📧 ГОТОВОЕ СООБЩЕНИЕ ДЛЯ РАССЫЛКИ:")
        print("="*60)
        print(result["message"])
        print("\n" + "="*60)
        print("📊 ИЗВЛЕЧЕННЫЕ ДАННЫЕ:")
        print("="*60)
        print(json.dumps(result["extracted_info"], ensure_ascii=False, indent=2))

    # ПРИМЕР 2: Несколько сайтов с дополнительными источниками
    # urls_with_sources = [
    #     {
    #         "main_url": "https://tau24.kz",
    #         "additional_urls": [
    #             "https://2gis.kz/almaty/firm/70000001018489394"
    #         ]
    #     },
    #     {
    #         "main_url": "https://example-resort.com",
    #         "additional_urls": [
    #             "https://2gis.kz/almaty/firm/123456789",
    #             "https://instagram.com/example_resort"
    #         ]
    #     }
    # ]
    # results = summarizer.batch_generate_messages(urls_with_sources)

    # ПРИМЕР 3: Один сайт БЕЗ дополнительных источников (старый вариант)
    # result = summarizer.generate_outreach_message("https://tau24.kz")


        for pattern, type_name in type_patterns:
            if re.search(pattern, text[:2000], re.IGNORECASE):
                info["type"] = type_name
                print(f"  ✓ type: {type_name}")
                break

        # Ищем локацию - казахстанские города
        cities = [
            'Алмат[ыи]', 'Астан[аы]', 'Шымкент', 'Караганд[аы]',
            'Актобе', 'Тараз', 'Павлодар', 'Усть-Каменогорск',
            'Семей', 'Атырау', 'Костанай', 'Кызылорд[аы]',
            'Боровое', 'Капчагай', 'Иссык', 'Талгар',
            'Медеу', 'Чимбулак', 'Кокшетау', 'Петропавловск'
        ]

        for city in cities:
            match = re.search(f'({city})', text[:3000], re.IGNORECASE)
            if match:
                info["location"] = match.group(1)
                print(f"  ✓ location: {match.group(1)}")
                break

        # Ищем соцсети
        social_patterns = [
            (r'instagram\.com|@[\w.]+|инстаграм', 'Instagram'),
            (r'facebook\.com|фейсбук', 'Facebook'),
            (r'vk\.com|вконтакте', 'VK'),
            (r'tiktok\.com|тикток', 'TikTok'),
            (r'youtube\.com|ютуб', 'YouTube'),
        ]

        for pattern, platform in social_patterns:
            if re.search(pattern, text[:5000], re.IGNORECASE):
                info["social_media"] = platform
                print(f"  ✓ social_media: {platform}")
                break

        # Ищем ключевые особенности
        features = []
        feature_patterns = [
            (r'(\d+)\s*(?:бассейн|pool)', 'бассейн'),
            (r'(\d+)\s*(?:сауна|sauna)', 'сауна'),
            (r'(\d+)\s*(?:джакузи|jacuzzi)', 'джакузи'),
            (r'ресторан|restaurant', 'ресторан'),
            (r'spa|спа', 'СПА'),
            (r'зоопарк|zoo', 'зоопарк'),
            (r'рыбалка|fishing', 'рыбалка'),
            (r'горн|mountain', 'горы'),
        ]

        for pattern, feature in feature_patterns:
            match = re.search(pattern, text[:3000], re.IGNORECASE)
            if match:
                if match.groups():
                    features.append(f"{match.group(1)} {feature}")
                else:
                    features.append(feature)

        if features:
            info["key_features"] = ", ".join(features[:3])
            print(f"  ✓ key_features: {info['key_features']}")

        return info

    def _translate_type_to_russian(self, type_text):
        """Переводит тип места на русский язык"""
        translations = {
            # Английские варианты
            'resort': 'курорт',
            'hotel': 'отель',
            'glamping': 'глэмпинг',
            'camping': 'кемпинг',
            'cottage': 'коттедж',
            'guesthouse': 'гостевой дом',
            'hostel': 'хостел',
            'motel': 'мотель',
            'villa': 'вилла',
            'apartment': 'апартаменты',
            'recreation center': 'база отдыха',
            'recreation complex': 'комплекс отдыха',
            'spa resort': 'спа-курорт',
            'wellness resort': 'оздоровительный курорт',
            'mountain resort': 'горный курорт',
            'eco-resort': 'эко-курорт',
            'family resort': 'семейный курорт',
            'sanatorium': 'санаторий',
            'rest house': 'дом отдыха',
            'tourist base': 'турбаза',
            'lodge': 'усадьба',
            'chalet': 'шале',
            'bungalow': 'бунгало',
            'cabin': 'домик',

            # Смешанные варианты
            'spa center': 'спа-центр',
            'wellness center': 'оздоровительный центр',
            'recreation area': 'зона отдыха',
            'holiday resort': 'курорт',
            'vacation resort': 'курорт',
        }

        # Приводим к нижнему регистру для поиска
        type_lower = type_text.lower().strip()

        # Проверяем точное совпадение
        if type_lower in translations:
            return translations[type_lower]

        # Проверяем частичное совпадение
        for eng, rus in translations.items():
            if eng in type_lower:
                return rus

        # Если уже на русском или не нашли перевод, возвращаем как есть
        return type_text

    def fetch_website(self, url):
        """Загружает содержимое веб-сайта"""
        try:
            headers = {
                'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
            }
            response = requests.get(url, headers=headers, timeout=10)
            response.raise_for_status()
            return response.text
        except Exception as e:
            return f"Ошибка при загрузке сайта: {str(e)}"

    def extract_text(self, html_content):
        """Извлекает текст из HTML"""
        soup = BeautifulSoup(html_content, 'html.parser')

        # Пытаемся найти название из title или h1
        title = ""
        if soup.title:
            title = soup.title.string

        # Удаляем ненужные элементы
        for tag in soup(['script', 'style', 'nav', 'footer', 'header']):
            tag.decompose()

        # Извлекаем текст
        text = soup.get_text(separator=' ', strip=True)

        # Очищаем от лишних пробелов
        text = ' '.join(text.split())

        # Ограничиваем длину для оптимальной работы LLM
        return text[:8000], title

    def extract_business_info(self, url, additional_urls=None):
        """Извлекает информацию о бизнесе для персонализации

        Args:
            url: основной URL сайта
            additional_urls: список дополнительных URL (например, 2ГИС, социальные сети)
        """

        # 1. Загружаем основной сайт
        print(f"Загрузка основного сайта: {url}")
        html_content = self.fetch_website(url)

        if html_content.startswith("Ошибка"):
            return {"error": html_content}

        # 2. Извлекаем текст с основного сайта
        print("Извлечение информации с основного сайта...")
        text_content, title = self.extract_text(html_content)

        # 3. Загружаем дополнительные источники
        additional_text = ""
        if additional_urls:
            print(f"\nЗагрузка дополнительных источников ({len(additional_urls)})...")
            for i, add_url in enumerate(additional_urls, 1):
                print(f"  {i}. {add_url}")
                add_html = self.fetch_website(add_url)
                if not add_html.startswith("Ошибка"):
                    add_text, _ = self.extract_text(add_html)
                    additional_text += f"\n\n--- Источник {i} ({add_url}) ---\n{add_text[:3000]}"

        # Объединяем всю информацию
        combined_text = text_content + additional_text

        print(f"\n📝 Общий объем текста для анализа: {len(combined_text)} символов")
        print(f"   - Основной сайт: {len(text_content)} символов")
        if additional_text:
            print(f"   - Доп. источники: {len(additional_text)} символов")

        # 4. Создаем улучшенный промпт с примерами
        prompt = f"""Extract business information from the text and format as JSON.

Text to analyze:
{combined_text[:10000]}

Extract these fields:
- name: business name
- type: type of establishment (hotel, resort, glamping, etc.)
- location: city or region
- social_media: social media platform mentioned
- key_features: main amenities

Respond ONLY with JSON in this exact format:
{{"name": "extracted name", "type": "extracted type", "location": "extracted location", "social_media": "extracted platform or неизвестно", "key_features": "extracted features"}}

JSON:"""

        # 5. Отправляем в Ollama с adjusted параметрами
        print(f"\nАнализ данных с моделью {self.model_name}...")

        # Пробуем несколько раз с разными настройками если не получается
        max_attempts = 2
        for attempt in range(max_attempts):
            try:
                if attempt > 0:
                    print(f"\n🔄 Попытка #{attempt + 1}...")

                response = ollama.generate(
                    model=self.model_name,
                    prompt=prompt,
                    options={
                        'temperature': 0.1 if attempt == 0 else 0.3,
                        'top_p': 0.9,
                        'num_predict': 600,
                    }
                )

                # Извлекаем ответ
                response_text = response['response'].strip()

                if attempt == 0:  # Показываем только в первый раз
                    print(f"\n🔍 Ответ модели (первые 500 символов):")
                    print("=" * 60)
                    print(response_text[:500])
                    print("=" * 60)

                # Очищаем ответ
                response_text = re.sub(r'```json\s*', '', response_text)
                response_text = re.sub(r'```\s*', '', response_text)
                response_text = response_text.strip()

                # Ищем JSON
                json_match = re.search(r'\{[^\{\}]*\}', response_text)

                if json_match:
                    json_str = json_match.group()

                    # Пытаемся распарсить
                    try:
                        parsed = json.loads(json_str)
                        # Проверяем что есть хотя бы одно нужное поле
                        if any(key in parsed for key in ["name", "type", "location"]):
                            print(f"✅ JSON успешно распарсен!")

                            # Дополняем недостающие поля
                            info = {
                                "name": parsed.get("name", "неизвестно"),
                                "type": parsed.get("type", "место отдыха"),
                                "location": parsed.get("location", "неизвестно"),
                                "social_media": parsed.get("social_media", "неизвестно"),
                                "key_features": parsed.get("key_features", "неизвестно")
                            }

                            print(f"\n📊 Извлеченные данные:")
                            for key, value in info.items():
                                print(f"  • {key}: {value}")

                            return info
                    except json.JSONDecodeError:
                        pass

                # Если JSON не найден или не распарсился, пробуем fallback
                if attempt == max_attempts - 1:
                    print(f"\n⚠️ JSON не найден или поврежден, использую умное извлечение...")
                    info = self._smart_extract_from_text(combined_text, title, response_text)

                    print(f"\n📊 Извлеченные данные:")
                    for key, value in info.items():
                        print(f"  • {key}: {value}")

                    return info

            except Exception as e:
                print(f"❌ Ошибка в попытке #{attempt + 1}: {str(e)}")
                if attempt == max_attempts - 1:
                    print(f"\n🔧 Использую аварийное извлечение...")
                    return self._smart_extract_from_text(combined_text, title, "")

    def generate_outreach_message(self, url, additional_urls=None, template=None):
        """Генерирует персонализированное сообщение рассылки

        Args:
            url: основной URL сайта
            additional_urls: список дополнительных URL (2ГИС, соцсети и т.д.)
            template: шаблон сообщения
        """

        if template is None:
            template = """Привет, {name}!

Мы заметили ваш потрясающий {type} рядом с {location} в {social_media}.
Ваше место явно заслуживает большей аудитории туристов!
mytravel.kz — это топовая платформа путешествий в Казахстане, где вы получите:
✨ 10K+ активных туристов ежемесячно
✨ Лучший рейтинг для премиум объектов
✨ Инструменты управления бронированиями

Хотим добавить вас в каталог. Можно поговорим?

Спасибо,
Команда mytravel.kz"""

        # Извлекаем информацию из всех источников
        info = self.extract_business_info(url, additional_urls)

        if "error" in info:
            return info["error"]

        # Переводим тип на русский
        type_translated = self._translate_type_to_russian(info.get("type", "место отдыха"))

        # Заполняем шаблон
        message = template.format(
            name=info.get("name", "[Название]"),
            type=type_translated,
            location=info.get("location", "[Локация]"),
            social_media=info.get("social_media", "социальных сетях")
        )

        # Сохраняем переведенный тип в info
        info["type_translated"] = type_translated

        return {
            "message": message,
            "extracted_info": info
        }

    def batch_generate_messages(self, urls_with_sources, template=None, output_file="outreach_messages.json"):
        """Генерирует сообщения для списка URL с дополнительными источниками

        Args:
            urls_with_sources: список словарей вида:
                [
                    {
                        "main_url": "https://site.com",
                        "additional_urls": ["https://2gis.kz/...", "https://instagram.com/..."]
                    },
                    ...
                ]
                или просто список URL строк (старый формат)
            template: шаблон сообщения
            output_file: имя файла для сохранения результатов
        """
        results = []

        for i, url_data in enumerate(urls_with_sources, 1):
            print(f"\n{'='*60}")

            # Поддержка старого формата (просто строка URL)
            if isinstance(url_data, str):
                main_url = url_data
                additional_urls = None
                print(f"Обработка {i}/{len(urls_with_sources)}: {main_url}")
            else:
                main_url = url_data.get("main_url")
                additional_urls = url_data.get("additional_urls", [])
                print(f"Обработка {i}/{len(urls_with_sources)}: {main_url}")
                if additional_urls:
                    print(f"  + {len(additional_urls)} дополнительных источника(ов)")

            print('='*60)

            result = self.generate_outreach_message(main_url, additional_urls, template)

            if isinstance(result, dict) and "message" in result:
                results.append({
                    "main_url": main_url,
                    "additional_urls": additional_urls if additional_urls else [],
                    "message": result["message"],
                    "info": result["extracted_info"]
                })

                print("\n📧 СГЕНЕРИРОВАННОЕ СООБЩЕНИЕ:")
                print("-" * 60)
                print(result["message"])
                print("-" * 60)
                print(f"\n📊 Извлеченная информация:")
                for key, value in result["extracted_info"].items():
                    print(f"  • {key}: {value}")
            else:
                results.append({
                    "main_url": main_url,
                    "additional_urls": additional_urls if additional_urls else [],
                    "error": result
                })
                print(f"\n❌ Ошибка: {result}")

        # Сохраняем результаты
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(results, f, ensure_ascii=False, indent=2)

        print(f"\n✅ Результаты сохранены в {output_file}")
        return results


# Пример использования
if __name__ == "__main__":
    # Инициализация
    summarizer = WebsiteSummarizer(model_name="qwen3:14b")

    # ПРИМЕР 1: Один сайт с дополнительными источниками
    print("🚀 ГЕНЕРАЦИЯ ПЕРСОНАЛИЗИРОВАННОГО СООБЩЕНИЯ\n")

    main_url = "https://tau24.kz"
    additional_urls = [
        "https://2gis.kz/almaty/firm/70000001018489394",  # 2ГИС
        # "https://instagram.com/tau_dastarkhan",  # Instagram (опционально)
    ]

    result = summarizer.generate_outreach_message(
        url=main_url,
        additional_urls=additional_urls
    )

    if isinstance(result, dict) and "message" in result:
        print("\n" + "="*60)
        print("📧 ГОТОВОЕ СООБЩЕНИЕ ДЛЯ РАССЫЛКИ:")
        print("="*60)
        print(result["message"])
        print("\n" + "="*60)
        print("📊 ИЗВЛЕЧЕННЫЕ ДАННЫЕ:")
        print("="*60)
        print(json.dumps(result["extracted_info"], ensure_ascii=False, indent=2))

    # ПРИМЕР 2: Несколько сайтов с дополнительными источниками
    # urls_with_sources = [
    #     {
    #         "main_url": "https://tau24.kz",
    #         "additional_urls": [
    #             "https://2gis.kz/almaty/firm/70000001018489394"
    #         ]
    #     },
    #     {
    #         "main_url": "https://example-resort.com",
    #         "additional_urls": [
    #             "https://2gis.kz/almaty/firm/123456789",
    #             "https://instagram.com/example_resort"
    #         ]
    #     }
    # ]
    # results = summarizer.batch_generate_messages(urls_with_sources)

    # ПРИМЕР 3: Один сайт БЕЗ дополнительных источников (старый вариант)
    # result = summarizer.generate_outreach_message("https://tau24.kz")
````
